```python
df_nav = pd.read_csv(
    "../data/processed/clean_nav.csv"
)

df_nav["date"] = pd.to_datetime(df_nav["date"])

dates = df_nav["date"].drop_duplicates().sort_values()

df_date = pd.DataFrame({
    "full_date": dates
})

df_date["date_key"] = range(1, len(df_date) + 1)
df_date["year"] = df_date["full_date"].dt.year
df_date["quarter"] = df_date["full_date"].dt.quarter
df_date["month"] = df_date["full_date"].dt.month
df_date["month_name"] = df_date["full_date"].dt.month_name()
df_date["day"] = df_date["full_date"].dt.day
df_date["day_name"] = df_date["full_date"].dt.day_name()

df_date.to_sql(
    "dim_date",
    engine,
    if_exists="append",
    index=False
)

print("dim_date loaded:", len(df_date))
```

```python
df_nav = pd.read_csv(
    "../data/processed/clean_nav.csv"
)

df_nav["date"] = pd.to_datetime(df_nav["date"])

df_nav = df_nav.sort_values(
    ["amfi_code", "date"]
)

df_nav["daily_return"] = (
    df_nav.groupby("amfi_code")["nav"].pct_change()
)
```

```python
df_date_db = pd.read_sql(
    "SELECT date_key, full_date FROM dim_date",
    engine
)

df_date_db["full_date"] = pd.to_datetime(
    df_date_db["full_date"]
)

df_nav = df_nav.merge(
    df_date_db,
    left_on="date",
    right_on="full_date",
    how="left"
)
```

```python
df_nav_fact = df_nav[
    [
        "amfi_code",
        "date_key",
        "date",
        "nav",
        "daily_return"
    ]
].copy()

df_nav_fact = df_nav_fact.rename(
    columns={"date": "nav_date"}
)

df_nav_fact["amfi_code"] = (
    df_nav_fact["amfi_code"].astype(str)
)

df_nav_fact.to_sql(
    "fact_nav",
    engine,
    if_exists="append",
    index=False
)

print("fact_nav loaded:", len(df_nav_fact))
```

```python
# Get dates from transactions
transaction_dates = df_trans["transaction_date"].dropna()

# Get dates from AUM
aum_dates = df_aum["date"].dropna()

# Combine all dates
all_dates = pd.concat(
    [df_nav["date"], transaction_dates, aum_dates]
).drop_duplicates().sort_values()

print("Total unique dates needed:", len(all_dates))
```

```python
existing_dates = pd.read_sql(
    "SELECT full_date FROM dim_date",
    engine
)

existing_dates["full_date"] = pd.to_datetime(
    existing_dates["full_date"]
)

missing_dates = all_dates[
    ~all_dates.isin(existing_dates["full_date"])
].sort_values()

print("New dates to add:", len(missing_dates))
```

```python
new_dates = pd.DataFrame({
    "full_date": missing_dates
})

max_key = pd.read_sql(
    "SELECT MAX(date_key) AS max_key FROM dim_date",
    engine
).iloc[0]["max_key"]

new_dates["date_key"] = range(
    int(max_key) + 1,
    int(max_key) + 1 + len(new_dates)
)

new_dates["year"] = new_dates["full_date"].dt.year
new_dates["quarter"] = new_dates["full_date"].dt.quarter
new_dates["month"] = new_dates["full_date"].dt.month
new_dates["month_name"] = new_dates["full_date"].dt.month_name()
new_dates["day"] = new_dates["full_date"].dt.day
new_dates["day_name"] = new_dates["full_date"].dt.day_name()
```

```python
query1 = """
SELECT
    amfi_code,
    scheme_name,
    aum_crore
FROM fact_performance
ORDER BY aum_crore DESC
LIMIT 5;
"""

result1 = pd.read_sql(query1, engine)
result1
```

```python
query2 = """
SELECT
    strftime('%Y-%m', nav_date) AS month,
    ROUND(AVG(nav), 2) AS average_nav
FROM fact_nav
GROUP BY strftime('%Y-%m', nav_date)
ORDER BY month;
"""

result2 = pd.read_sql(query2, engine)
result2
```

```python
query3 = """
WITH yearly_sip AS (
    SELECT
        strftime('%Y', month) AS year,
        SUM(sip_inflow_crore) AS total_sip_inflow
    FROM monthly_sip_inflows
    GROUP BY strftime('%Y', month)
)

SELECT
    year,
    ROUND(total_sip_inflow, 2) AS total_sip_inflow,
    ROUND(
        (total_sip_inflow - LAG(total_sip_inflow) OVER (ORDER BY year))
        * 100.0
        / LAG(total_sip_inflow) OVER (ORDER BY year),
        2
    ) AS yoy_growth_pct
FROM yearly_sip
ORDER BY year;
"""

result3 = pd.read_sql(query3, engine)
result3
```

```python
query4 = """
SELECT
    state,
    COUNT(*) AS transaction_count,
    ROUND(SUM(amount_inr), 2) AS total_amount
FROM fact_transactions
GROUP BY state
ORDER BY transaction_count DESC;
"""

result4 = pd.read_sql(query4, engine)
result4
```

```python
query5 = """
SELECT
    amfi_code,
    scheme_name,
    expense_ratio_pct
FROM dim_fund
WHERE expense_ratio_pct < 1
ORDER BY expense_ratio_pct;
"""

result5 = pd.read_sql(query5, engine)
result5
```

```python
query6 = """
SELECT
    transaction_type,
    COUNT(*) AS transaction_count,
    ROUND(AVG(amount_inr), 2) AS average_amount
FROM fact_transactions
GROUP BY transaction_type
ORDER BY average_amount DESC;
"""

result6 = pd.read_sql(query6, engine)
result6
```

```python
query7 = """
SELECT
    amfi_code,
    ROUND(AVG(nav), 2) AS average_nav
FROM fact_nav
GROUP BY amfi_code
ORDER BY average_nav DESC
LIMIT 10;
"""

result7 = pd.read_sql(query7, engine)
result7
```

```python
query8 = """
SELECT
    risk_category,
    COUNT(*) AS fund_count
FROM dim_fund
GROUP BY risk_category
ORDER BY fund_count DESC;
"""

result8 = pd.read_sql(query8, engine)
result8
```

```python
query9 = """
SELECT
    city_tier,
    COUNT(*) AS transaction_count,
    ROUND(SUM(amount_inr), 2) AS total_amount
FROM fact_transactions
GROUP BY city_tier
ORDER BY total_amount DESC;
"""

result9 = pd.read_sql(query9, engine)
result9
```

```python
query10 = """
SELECT
    fund_house,
    ROUND(SUM(aum_crore), 2) AS total_aum_crore
FROM fact_aum
GROUP BY fund_house
ORDER BY total_aum_crore DESC;
"""

result10 = pd.read_sql(query10, engine)
result10
```
