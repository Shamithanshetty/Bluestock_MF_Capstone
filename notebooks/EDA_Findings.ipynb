# EDA Findings

* NAV Trend: The daily NAV analysis of 40 mutual fund schemes from 2022–2026 shows an overall upward movement, with a post-COVID recovery period, a strong 2023 rally, and corrections and volatility during 2024.
  Supporting Charts: NAV Trend Lines (Matplotlib) and NAV Trend Lines (Plotly)

* AUM Growth: AUM increased across the analyzed fund houses between 2022 and 2025, with SBI Mutual Fund recording the highest absolute AUM of approximately ₹12.50 lakh crore in 2025.
  Supporting Chart: AUM Growth by AMC

* SIP Inflow Growth: Monthly SIP inflows show a strong increasing trend from 2022 to 2025, reaching the ₹31,002 crore milestone in December 2025.
  Supporting Chart: SIP Inflow Trend

* Category-wise Inflows: The category-wise heatmap shows significant variation in monthly net inflows across fund categories, indicating changing investor preferences over time.
  Supporting Chart: Category Heatmap

* Investor Age Distribution: Investors in the 26–35 age group form the largest transaction segment, with 13,463 transactions in the analyzed dataset.
  Supporting Chart: Demographics_investor 

* SIP Amount by Age: SIP investment amounts vary across age groups, showing differences in investment behavior and contribution levels among investor segments.
  Supporting Chart: Demographics_sip 

* Geographic SIP Distribution: SIP investment amounts vary considerably across states, while T30 cities account for a larger share of SIP transactions than B30 cities.
  Supporting Charts: Geo Distribution_sip_State and Geo Distribution_t30 vs_b30 

* Folio Growth: Total mutual fund folios increased from approximately 13.26 crore in January 2022 to 26.12 crore in December 2025, indicating substantial growth in investor participation.
  Supporting Chart: folio_count_growth

* NAV Return Correlation: The correlation matrix shows different levels of co-movement among the daily NAV returns of the 10 selected funds, providing insights into similarities in fund return behavior and diversification potential.
  Supporting Chart: Correlation_Matrix

* Sector Allocation: The holdings of the analyzed equity funds are distributed across 14 sectors, showing diversified exposure across multiple industries.
  Supporting Chart: Sector_Allocation
