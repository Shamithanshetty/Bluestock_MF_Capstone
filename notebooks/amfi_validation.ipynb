```python
# Day1: Validate AMFI codes — check all codes in fund_master exist in nav_history
```

```python
import pandas as pd
```

```python
fund_master = pd.read_csv("../data/raw/01_fund_master.csv")
nav_history = pd.read_csv("../data/raw/02_nav_history.csv")
```

```python
print("Fund Master AMFI codes:", fund_master["amfi_code"].nunique())
print("NAV History AMFI codes:", nav_history["amfi_code"].nunique())
```

```python
missing_codes = set(fund_master["amfi_code"]) - set(nav_history["amfi_code"])

print("Missing AMFI codes:", missing_codes)
print("Number of missing codes:", len(missing_codes))
```

```python
validation = pd.merge(
    fund_master[["amfi_code", "scheme_name"]],
    nav_history[["amfi_code"]],
    on="amfi_code",
    how="left",
    indicator=True
)

print(validation["_merge"].value_counts())
```
